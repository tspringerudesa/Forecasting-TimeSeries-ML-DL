## Activation functions

The functions share one input range so saturation, boundedness, sparsity, negative slope, and smoothness are directly comparable.

## Recurrent family

The figure compresses the vanilla RNN, LSTM, and GRU into comparable information-flow cells. It emphasizes their state and gating differences without attempting to replace the full equations in the paper.

```python
fig, path = recurrent_family_figure()
print(path)
plt.show()
```

## Temporal convolution

A one-dimensional kernel is shown sliding across adjacent time steps. Weight sharing produces a temporal feature map, while stacking and dilation enlarge the receptive field.

```python
fig, path = convolution_figure()
print(path)
plt.show()
```

## MLP and KAN comparison

This is an original redraw of the conceptual distinction introduced by Liu et al.: MLPs place fixed nonlinearities at nodes and scalar weights on edges, whereas KANs place learnable univariate functions on edges and use additive nodes.

```python
fig, path = kan_vs_mlp_figure()
print(path)
plt.show()
```

## Scaled dot-product attention

The left panel follows the matrix operations in the Transformer equation. The right panel illustrates how every query position receives a probability distribution over key positions.

```python
fig, path = attention_figure()
print(path)
plt.show()
```

## Validation-strategy summary

The final figure consolidates the Chapter 20 strategies: basic holdout, expanding and rolling Rep-Holdout-O, their gap variants, blocked cross-validation, and blocked cross-validation with gaps. All blocks preserve the displayed chronological order.

```python
fig, path = validation_strategies_figure()
print(path)
plt.show()
```
